```python
for column in df.select_dtypes(include="object").columns:
    spaces = df[column].astype(str).str.contains(r"^\s|\s$", regex=True).sum()
    print(f"{column}: {spaces} values with leading/trailing spaces")
```

```python
df_raw = df.copy()

print("Original shape:", df_raw.shape)
print("Working shape:", df.shape)
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)

print(df.columns.tolist())
```

```python
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()

print("Text columns cleaned.")
```

```python
for column in text_columns:
    print(column, df[column].isna().sum())
```

```python
print("Values above 10 hours:", (df["avg_watch_time_per_day"] > 10).sum())

print("Values above 24 hours:", (df["avg_watch_time_per_day"] > 24).sum())

print("\nTop 10 values:")
print(
    df["avg_watch_time_per_day"]
    .sort_values(ascending=False)
    .head(10)
)
```

```python
df[["watch_hours", "avg_watch_time_per_day"]].head(20)
df["watch_hours_to_avg_ratio"] = (
    df["watch_hours"] / df["avg_watch_time_per_day"]
)

print(df["watch_hours_to_avg_ratio"].describe())
```

```python
df.loc[
    df["avg_watch_time_per_day"] > 10,
    ["watch_hours", "last_login_days", "avg_watch_time_per_day"]
].head(20)
```

```python
invalid_watch_time = df[df["avg_watch_time_per_day"] > 24]

print("Invalid records:", len(invalid_watch_time))
print("\nInvalid records:")
print(
    invalid_watch_time[
        [
            "customer_id",
            "watch_hours",
            "last_login_days",
            "avg_watch_time_per_day",
            "churned"
        ]
    ]
)
```

```python
duplicate_ids = df[df["customer_id"].duplicated(keep=False)]

print("Duplicate customer IDs:", duplicate_ids["customer_id"].nunique())
print("Rows with duplicate IDs:", len(duplicate_ids))
```

```python
missing_values = df.isnull().sum()

print(missing_values)
print("\nTotal missing values:", missing_values.sum())
```

```python
if "watch_time_group" in df.columns:
    df.drop(columns=["watch_time_group"], inplace=True)

print(df.columns.tolist())
print("Shape:", df.shape)
```

```python
print("Original shape :", df_raw.shape)
print("Cleaned shape  :", df.shape)

print("\nColumns changed:")
print(df_raw.columns.equals(df.columns))

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nDuplicate customer IDs:")
print(df["customer_id"].duplicated().sum())
```
